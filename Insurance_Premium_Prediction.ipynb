```python
import pandas as pd, numpy as np
df = pd.read_csv('insurance_working_dataset.csv')
df.head()
```

```python
print(df.shape)
print(df.isnull().sum())
df.describe()
```

```python
import matplotlib.pyplot as plt
plt.hist(df['charges'], bins=30)
plt.title('Distribution of Insurance Charges')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
X=df.drop('charges',axis=1); y=df['charges']
pre=ColumnTransformer([('cat',OneHotEncoder(handle_unknown='ignore'),['sex','smoker','region']),('num','passthrough',['age','bmi','children'])])
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42)
model=Pipeline([('preprocessor',pre),('model',GradientBoostingRegressor(random_state=42))])
model.fit(X_train,y_train)
p=model.predict(X_test)
print('MAE',mean_absolute_error(y_test,p))
print('RMSE',np.sqrt(mean_squared_error(y_test,p)))
print('R2',r2_score(y_test,p))
```

```python
new_customer=pd.DataFrame({'age':[25],'sex':['male'],'bmi':[24.5],'children':[0],'smoker':['no'],'region':['northeast']})
print('Predicted charge:',model.predict(new_customer)[0])
```
